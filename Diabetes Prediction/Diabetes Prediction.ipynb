<spam style="font-size:40px;">**Importing the Required Libraries**</spam>

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn import svm
```

<spam style="font-size:40px;">**Loading the Diabetes Dataset**</spam>

```python
diabetes_df=pd.read_csv("diabetes.csv")
```

```python
#splitting the columns 
#independent or features column as x
x=diabetes_df.drop(columns = "Outcome",axis=1)
#dependent or target column as y
y=diabetes_df['Outcome']
```

```python
#displaying first 5 records of x
x.head()
```

```python
#displaying first 5 records of y
y.head()
```

<spam style="font-size:20px;">**Standardizing the features**</spam>

```python
scaler = StandardScaler()
standardized_data = scaler.fit_transform(x)
standardized_data
```

```python
x= standardized_data
y=diabetes_df['Outcome']
```

<spam style="font-size:20px;">**Splitting the Training data**</spam>

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,stratify=y,random_state=2)
```

```python
print(x.shape)
print(x_train.shape)
print(x_test.shape)
```

<spam style="font-size:20px;">**Training the Model**</spam>

```python
Classifier = svm.SVC(kernel='linear')
```

```python
Classifier.fit(x_train,y_train)
```

<spam style="font-size:40px;">**Model Evaluation**</spam>

```python
x_train_data_prediction = Classifier.predict(x_train)
x_train_accuracy = accuracy_score(x_train_data_prediction,y_train)
print(" Accuracy score of  training data:",x_train_accuracy)
```

```python
x_test_data_prediction = Classifier.predict(x_test)
x_test_accuracy = accuracy_score(x_test_data_prediction,y_test)
print(" Accuracy score of  test data:",x_test_accuracy)
```

<spam style="font-size:40px;">**Building Predictive System**</spam>

```python
input_data = (5,166,72,19,175,25.8,0.587,50)

#changing input data into array
input_as_array = np.asarray(input_data)

#reshaping the array as we are predicting for one instance
input_data_reshape=input_as_array.reshape(1,-1)

#standardizing the input data
std_input = scaler.transform(input_data_reshape)

#predicting the input data
Prediction =Classifier.predict(std_input)
print(Prediction)

if(Prediction ==0):
    print("patient has no diabetes")
else:
    print("patient has diabetes")
```
